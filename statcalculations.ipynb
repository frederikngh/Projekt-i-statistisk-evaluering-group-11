# McNemars test for the paired questions. Questions presented as text/tables written out vs questions presented as a screenshot

```python
# Import library
from statsmodels.stats.contingency_tables import mcnemar

# Create a dataset
data = [[45, 9],
        [35, 38]]

print(mcnemar(data, exact=False, correction=False))
```

# Two-proportion z-test between accuracy of questions with only text and questions with only screenshot 

```python
from statsmodels.stats.proportion import proportions_ztest
z, p = proportions_ztest([98, 52], [134, 144])
print(z)
print(p)
```

## one-sided two-proportion z-test between accuracy before knowledge cutoff and after knowledge cutoff

```python
from statsmodels.stats.proportion import proportions_ztest
z, p = proportions_ztest([263, 46], [498, 68], alternative='larger')
print(z)
print(p)
```

# 95% confidence intervals for the accuracies

```python
from statsmodels.stats.proportion import proportion_confint
print(f"CI text only = [{round(proportion_confint(98, 134)[0]*100,2)}%, {round(proportion_confint(98, 134)[1]*100,2)}%]")
print(f"CI paired text = [{round(proportion_confint(80, 127)[0]*100,2)}%, {round(proportion_confint(80, 127)[1]*100,2)}%]")
print(f"CI paired image = [{round(proportion_confint(54, 127)[0]*100,2)}%, {round(proportion_confint(54, 127)[1]*100,2)}%]")
print(f"CI image only = [{round(proportion_confint(52, 144)[0]*100,2)}%, {round(proportion_confint(52, 144)[1]*100,2)}%]")
print(f"CI overall accuracy = [{round(proportion_confint(309, 566)[0]*100,2)}%, {round(proportion_confint(309, 566)[1]*100,2)}%]")
```

```python
import numpy as np

#paired confidence interval for the difference of the paired questions
b = 35
c = 9
n = 127
diff = (b-c)/n
Standard_error = np.sqrt(b+c-(b-c)**2/n)/n
print(f"paired: text - image = {round(100*diff,1)} pts  95% CI [{round(100*(diff-1.96*Standard_error),1)}, {round(100*(diff+1.96*Standard_error),1)}]")
```

```python
from statsmodels.stats.proportion import confint_proportions_2indep
# not paired questions
print(f"not paired: text - image = {confint_proportions_2indep(98, 134, 52, 144, method='wald')}")
```
