# McNemars test for the paired questions. Questions presented as text/tables written out vs questions presented as a screenshot

```python
# Import library
from statsmodels.stats.contingency_tables import mcnemar

# Create a dataset
data = [[45, 9],
        [35, 38]]

print(mcnemar(data, exact=True))
```

# Binomial test to check if all question types are better than random guessing

```python
from scipy.stats import binomtest
print(f"pvalue for text only = {binomtest(98, 134, 0.25, alternative="greater").pvalue}")
print(f"pvalue for screenshot only = {binomtest(52, 144, 0.25, alternative="greater").pvalue}")
print(f"pvalue paired, text = {binomtest(80, 127, 0.25, alternative="greater").pvalue}")
print(f"pvalue paired, image = {binomtest(54, 127, 0.25, alternative="greater").pvalue}")
```

# Two-proportion z-test between accuracy of questions with only text and questions with only screenshot 

```python
from statsmodels.stats.proportion import proportions_ztest
z, p = proportions_ztest([98, 52], [134, 144])
print(z)
print(p)
```

## one-sided two-proportion z-test between accuracy before knowledge cutoff and after knowledge cutoff

```python
from statsmodels.stats.proportion import proportions_ztest
z, p = proportions_ztest([263, 46], [498, 68], alternative='larger')
print(z)
print(p)
```

# 95% confidence intervals (normal approximation)
One statsmodels call per accuracy. The unpaired text-vs-graph difference uses the standard two-proportion CI. The paired text-vs-image difference needs the paired formula: the two arms are the SAME 127 questions, so they are correlated and the independent-groups CI would be wrong (too wide).

```python
from statsmodels.stats.proportion import proportion_confint
print(f"CI pure text = {proportion_confint(98, 134)}")
print(f"CI paired text = {proportion_confint(80, 127)}")
print(f"CI paired image = {proportion_confint(54, 127)}")
print(f"CI geometric image = {proportion_confint(52, 144)}")
```

```python
import numpy as np
# paired diff uses the discordant counts (35 only-text, 9 only-image); the two arms are correlated
b, c, n = 35, 9, 127
diff = (b - c) / n
se = np.sqrt(b + c - (b - c)**2 / n) / n
print(f"CI paired text - image = ({diff - 1.96*se}, {diff + 1.96*se})")
```

```python
from statsmodels.stats.proportion import confint_proportions_2indep
# independent groups (different questions), so the standard two-proportion CI applies
print(f"CI text - geometric = {confint_proportions_2indep(98, 134, 52, 144, method='wald')}")
```
